## **mesurer les anomalies**

```python
# les Statuts écrits de plusieurs façons
statuts_ok = ["livree", "annulee", "remboursee", "en_attente_paiement", "expediee", "payee"]
anormaux = commandes[~commandes["statut"].isin(statuts_ok)]
print("Statuts mal écrits :", len(anormaux))

# les Pays écrits de plusieurs façons
pays_ok = ["FR", "CH", "BE", "ES", "DE", "IT", "MA", "TN", "AE", "LU", "IL"]
anormaux = commandes[~commandes["pays_livraison"].isin(pays_ok)]
print("Pays mal écrits :", len(anormaux))

# les Commandes en francs suisses
anormaux = commandes[commandes["devise"] == "CHF"]
print("Commandes en CHF :", len(anormaux))

# les Montants HT négatifs ou nuls
anormaux = commandes[commandes["montant_ht"] <= 0]
print("Montants HT <= 0 :", len(anormaux))

# les Clients absents ou inexistants
anormaux = commandes[~commandes["id_client"].isin(clients["id_client"])]
print("Clients absents ou inexistants :", len(anormaux))

# les Lignes de commande avec un produit inexistant
anormaux = lignes_commande[~lignes_commande["id_produit"].isin(produits["id_produit"])]
print("Produits inexistants :", len(anormaux))

# les Lignes de commande invalides
l = lignes_commande
anormaux = l[(l["quantite"] <= 0) | (l["prix_unitaire_ht_eur"] <= 0) | (l["remise_pct"] < 0) | (l["remise_pct"] > 100)]
print("Lignes invalides :", len(anormaux))

# les Dates de commande illisibles
dates = pd.to_datetime(commandes["date_commande"], errors="coerce")
print("Dates illisibles :", dates.isna().sum())
print("Dates dans le futur :", (dates > pd.Timestamp.today()).sum())

# les Stock : doublons et valeurs manquantes
print("Relevés de stock en double :", stock_hebdo.duplicated(["id_produit", "date_releve"]).sum())
print("Stock manquant :", stock_hebdo["stock_unites"].isna().sum())
```

### **nettoyage**   

On va corrige que les anomalies qui faussent les analyses commerciales. Les autres ont va simplement les signalées.

 **préparer**  on travaille sur des copies (cmd, lig, stk). mes tableaux d’origine (commandes, lignes_commande, stock_hebdo)

```python
cmd = commandes.copy()
lig = lignes_commande.copy()
stk = stock_hebdo.copy()
bilan = []
```

```python
statuts_ok = ["livree", "annulee", "remboursee", "en_attente_paiement", "expediee", "payee"]
nb = (~cmd["statut"].isin(statuts_ok)).sum()

cmd["statut"] = cmd["statut"].str.strip().str.lower()
cmd["statut"] = cmd["statut"].replace({"livrée": "livree", "delivered": "livree", "annulée": "annulee",
    "cancelled": "annulee", "refunded": "remboursee", "paid": "payee", "payée": "payee",
    "pending": "en_attente_paiement"})

bilan.append(("commandes", "statuts harmonisés", nb, len(commandes), len(cmd)))
```

```python
pays_ok = ["FR", "CH", "BE", "ES", "DE", "IT", "MA", "TN", "AE", "LU", "IL"]
nb = (~cmd["pays_livraison"].isin(pays_ok)).sum()

cmd["pays_livraison"] = cmd["pays_livraison"].str.strip().str.upper()
cmd["pays_livraison"] = cmd["pays_livraison"].replace({"FRANCE": "FR", "ALLEMAGNE": "DE", "MAROC": "MA",
    "SUISSE": "CH", "ITALIE": "IT", "BELGIQUE": "BE", "LUXEMBOURG": "LU", "TUNISIE": "TN",
    "ÉMIRATS ARABES UNIS": "AE", "ESPAGNE": "ES", "ISRAËL": "IL"})

bilan.append(("commandes", "pays harmonisés (codes ISO)", nb, len(commandes), len(cmd)))
```

**dates de commande**

```python
essai = pd.to_datetime(cmd["date_commande"], errors="coerce")
nb = essai.isna().sum() + (essai > pd.Timestamp.today()).sum()

d1 = pd.to_datetime(cmd["date_commande"], format="%Y-%m-%d %H:%M:%S", errors="coerce")
d2 = pd.to_datetime(cmd["date_commande"], format="%d/%m/%Y %H:%M", errors="coerce")
cmd["date_commande"] = d1.fillna(d2)
cmd.loc[cmd["date_commande"] > pd.Timestamp.today(), "date_commande"] = pd.NaT

bilan.append(("commandes", "dates converties (20 restent vides)", nb, len(commandes), len(cmd)))
```

**montants négatifs** repère les commandes dont le montant est négatif ou nul et passe en valeur absolue (positive) les trois colonnes de montant (montant_ht, montant_tva, montant_ttc)

```python
negatifs = cmd["montant_ht"] <= 0
nb = negatifs.sum()

for col in ["montant_ht", "montant_tva", "montant_ttc"]:
    cmd.loc[negatifs, col] = cmd.loc[negatifs, col].abs()

bilan.append(("commandes", "montants négatifs passés en positif", nb, len(commandes), len(cmd)))
```

**lignes de commande invalides** supprime les lignes jugées invalides selon des règles métier sur la quantité, le prix unitaire et le pourcentage de remise.

```python
commande_invalides = ((lig["quantite"] <= 0) | (lig["prix_unitaire_ht_eur"] <= 0)
             | (lig["remise_pct"] < 0) | (lig["remise_pct"] > 100))
nb = commande_invalides.sum()
lig = lig[~commande_invalides]

bilan.append(("lignes_commande", "lignes invalides supprimées", nb, len(lignes_commande), len(lig)))
```

**doublons de stock** on va supprime les lignes strictement identiques (doublons).. on garder une seule

```python
nb = stk.duplicated().sum()
stk = stk.drop_duplicates()

bilan.append(("stock_hebdo", "doublons supprimés", nb, len(stock_hebdo), len(stk)))
```

**sauvegarde** (les bruts restent intacts)

Les nouveaux fichiers ont un autre nom que les fichiers bruts

```python
cmd.to_csv("commandes_propre.csv", index=False)
lig.to_csv("lignes_commande_propre.csv", index=False)
stk.to_csv("stock_hebdo_propre.csv", index=False)

pd.DataFrame(bilan, columns=["table", "traitement", "lignes concernées", "avant", "après"])
```

### **contrôles de qualité**

```python
def controle(nom, ok):
    print("OK     " if ok else "ERREUR ", nom)

#  Valeurs harmonisées
controle("Statuts et pays valides",
         cmd["statut"].isin(statuts_ok).all() and cmd["pays_livraison"].isin(pays_ok).all())

#  Valeurs impossibles disparues
controle("Plus de montant HT <= 0 ni de ligne invalide",
         (cmd["montant_ht"] <= 0).sum() == 0
         and ((lig["quantite"] <= 0) | (lig["prix_unitaire_ht_eur"] <= 0)
              | (lig["remise_pct"] < 0) | (lig["remise_pct"] > 100)).sum() == 0)

#  Pas de doublon
controle("Pas de doublon (commandes, lignes, stock)",
         cmd["id_commande"].is_unique and lig["id_ligne"].is_unique
         and stk.duplicated(["id_produit", "date_releve"]).sum() == 0)
```
